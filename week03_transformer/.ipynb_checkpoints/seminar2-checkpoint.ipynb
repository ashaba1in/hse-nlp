### Токенизация

Токенизация – процесс разбиения текста на подстроки (токены). Токенизация в первую очередь нужна для уменьшения размера словаря. При этом при токенизации мы хотим, чтобы все токены были максимально репрезентативными. Таким образом, мы балансируем между размером словаря и репрезентативностью токенов.

```python
from tokenizers import Tokenizer, models

tokenizer = Tokenizer(models.BPE())
```

Токенизация включает в себя несколько компонент:

1. **Нормализация** - предварительную очистку текста (приведение к нижнему регистру, замена unicode символов на ascii и т.д.)

```python
from tokenizers import normalizers
from tokenizers.normalizers import NFD, StripAccents, Lowercase, Strip

normalizer = normalizers.Sequence([NFD(), StripAccents(), Lowercase(), Strip()])
normalizer.normalize_str(" Héllò hôw are ü?")
```

```python
tokenizer.normalizer = normalizer
```

2. **Предварительная токенизация** — процесс разбиения текста на более мелкие фрагменты, задающие верхнюю границу того, какими будут токены в токенизации. Обычно предварительная токенизация разобивает текст на слова. Тогда итоговые токены будут частями этих слов.

```python
from tokenizers.pre_tokenizers import Whitespace

pre_tokenizer = Whitespace()
pre_tokenizer.pre_tokenize_str("Hello! How are you? I'm fine, thank you.")
```

Заметьте, что строка "I'm" разбилась на ["I", "'", "m"]. Это может быть не очень хорошо, учитывая, что "'m" – то же самое, что "am". Для более "правильного" разбиения можно использовать претокенизацию на основе правил. Самые популярные инструменты для этого – `spaCy` и `Moses`.

```python
from mosestokenizer import MosesTokenizer

tokenize = MosesTokenizer('en')
tokenize("Hello! How are you? I'm fine, thank you.")
```

```python
tokenizer.pre_tokenizer = pre_tokenizer
```

3. **Модель** – главная часть токенизации. Модель необходимо обучать на корпусе текстов, чтобы получить словарь, подходящий под конкретные данные. Она применяется к результату претокенизации. И ее задача – разбить "слова" на более мелкие составляющие согласно выученным правилам.

Наиболее распространенные виды моделей: `BPE`, `WordPiece`, `Unigram`.

#### Byte-Pair Encoding (BPE)

__Алгоритм обучения:__

1. Считаем, сколько раз каждое слово встречается в корпусе
2. Создаем словарь токенов, который пока что состоит из всех уникальных символов
3. Находим пару токенов, которая чаще всего встречается вместе. Склеиваем ее в токен и добавляем в словарь.
4. Повторяем 3, пока словарь не достигнет нужного размера.


__Пример:__

Корпус слов со встречаемостями: (`("hug", 2)`, `("pug", 1)`, `("pun", 4)`, `("bun", 3)`, `("hugs", 1)`)   
Максимальный размер словаря – __10__.

__Шаг 1.__    
Словарь: [`"b"`, `"g"`, `"h"`, `"n"`, `"p"`, `"s"`, `"u"`], размер: __7__.   
Корпус: [`("h" "u" "g", 3)`, `("p" "u" "g", 1)`, `("p" "u" "n", 3)`, `("b" "u" "n", 3)`, `("h" "u" "g" "s", 1)`]

Чаще всего встречается пара (`"u" "n"`) – 6 раз. Добавляем ее в словарь и обновляем корпус.

__Шаг 2.__    
Словарь: [`"b"`, `"g"`, `"h"`, `"n"`, `"p"`, `"s"`, `"u"`, `"un"`], размер: __8__.   
Корпус: [`("h" "u" "g", 3)`, `("p" "u" "g", 1)`, `("p" "un", 3)`, `("b" "un", 3)`, `("h" "u" "g" "s", 1)`]

Чаще всего встречается пара (`"u" "g"`) – 5 раз. Добавляем ее в словарь и обновляем корпус.

__Шаг 3.__    
Словарь: [`"b"`, `"g"`, `"h"`, `"n"`, `"p"`, `"s"`, `"u"`, `"un"`, `"ug"`], размер: __9__.    
Корпус: [`("h" "ug", 3)`, `("p" "ug", 1)`, `("p" "un", 3)`, `("b" "un", 3)`, `("h" "ug" "s", 1)`]

Чаще всего встречается пара (`"h" "ug"`) – 4 раза. Добавляем ее в словарь и обновляем корпус.

__Шаг 4.__    
Словарь: [`"b"`, `"g"`, `"h"`, `"n"`, `"p"`, `"s"`, `"u"`, `"un"`, `"ug"`, `"hug"`], размер: __10__.    
Корпус: [`("hug", 3)`, `("p" "ug", 1)`, `("p" "un", 3)`, `("b" "un", 3)`, `("hug" "s", 1)`]


__Процесс токенизации:__

1. Нормализация
1. Предварительная токенизация
1. Разделение слов на отдельные символы
1. Применение правил слияния (в порядке их появления в словаре) к разделенным словам

```python
def preprocess(texts):
    clean_texts = []
    for i in range(len(texts)):
        clean_texts.append(re.sub(r"[^\w\s]", "", texts[i]).lower())
    return np.array(clean_texts)

def print_texts(texts):
    for t in texts:
        print(t, end='\n\n')
```

```python
clean_texts = preprocess(texts)

print_texts(clean_texts[idxs])
```

```python
train_texts, test_texts = np.split(clean_texts, [int(0.8 * len(clean_texts))])
len(train_texts), len(test_texts)
```

```python
SPECIAL_TOKENS = dict(bos_token='[BOS]', eos_token='[EOS]', unk_token='[UNK]', pad_token='[PAD]')


class CharacterTokenizer:
    def __init__(
        self, corpus, newline_token,
        *args, **kwargs
    ):
        self.newline_token = newline_token

        # collect the vocabulary of tokens to get token → id mapping
        self.token2id = {tok: i for i, tok in enumerate(SPECIAL_TOKENS.values())}
        corpus_tokens = sum(self.encode(corpus)['tokens'], [])
        new_tokens = set(np.unique(corpus_tokens)) - set(self.token2id)
        self.token2id.update({tok: len(self.token2id) + i for i, tok in enumerate(new_tokens)})
        self.id2token = list(self.token2id)

        self.pad_token_id = self.token2id[SPECIAL_TOKENS['pad_token']]

    def encode(self, texts, max_length=None):
        tokenized_texts = defaultdict(list)
        for text in texts:
            tokens = [SPECIAL_TOKENS['bos_token']] + [t.replace('\n', self.newline_token) for t in list(text)]
            if max_length is not None:
                tokens = tokens[:max_length-1]
            tokens.append(SPECIAL_TOKENS['eos_token'])
            tokenized_texts['tokens'].append(tokens)
            tokenized_texts['input_ids'].append(torch.tensor([
                self.token2id.get(t, self.token2id[SPECIAL_TOKENS['unk_token']]) for t in tokens
            ]))

        return tokenized_texts

    def decode(self, input_ids, skip_special_tokens=True):
        texts = []
        for ids in input_ids:
            tokens = [self.id2token[i] for i in ids]
            if skip_special_tokens:
                tokens = [t for t in tokens if t not in SPECIAL_TOKENS.values()]
            texts.append(''.join(tokens).replace(self.newline_token, '\n'))
        return texts

    def __len__(self):
        return len(self.token2id)


class WordTokenizer:
    def __init__(
        self, corpus, newline_token,
        *args, **kwargs
    ):
        self.newline_token = newline_token

        # collect the vocabulary of tokens to get token → id mapping
        self.token2id = {tok: i for i, tok in enumerate(SPECIAL_TOKENS.values())}
        corpus_tokens = sum(self.encode(corpus)['tokens'], [])
        new_tokens = set(np.unique(corpus_tokens)) - set(self.token2id)
        self.token2id.update({tok: len(self.token2id) + i for i, tok in enumerate(new_tokens)})
        self.id2token = list(self.token2id)

        self.pad_token_id = self.token2id[SPECIAL_TOKENS['pad_token']]

    def encode(self, texts, max_length=None):
        tokenized_texts = defaultdict(list)
        for text in texts:
            tokens = [SPECIAL_TOKENS['bos_token']] + text.replace('\n', f' {self.newline_token} ').split()
            if max_length is not None:
                tokens = tokens[:max_length-1]
            tokens.append(SPECIAL_TOKENS['eos_token'])
            tokenized_texts['tokens'].append(tokens)
            tokenized_texts['input_ids'].append(torch.tensor([
                self.token2id.get(t, self.token2id[SPECIAL_TOKENS['unk_token']]) for t in tokens
            ]))

        return tokenized_texts

    def decode(self, input_ids, skip_special_tokens=True):
        texts = []
        for ids in input_ids:
            tokens = [self.id2token[i] for i in ids]
            if skip_special_tokens:
                tokens = [t for t in tokens if t not in SPECIAL_TOKENS.values()]
            texts.append(
                ' '.join(tokens).replace(self.newline_token, '\n').replace(' \n', '\n').replace('\n ', '\n')
            )
        return texts

    def __len__(self):
        return len(self.token2id)


class BPETokenizer:
    def __init__(
        self, corpus, newline_token, vocab_size=1024,
        *args, **kwargs
    ):
        self.newline_token = newline_token

        texts_bpe = [t.replace('\n', f' {newline_token} ') for t in corpus]
        self.tokenizer = Tokenizer(models.BPE())
        trainer = trainers.BpeTrainer(
            vocab_size=vocab_size, min_frequency=1,
            special_tokens=list(SPECIAL_TOKENS.values()),
            continuing_subword_prefix='##',
        )
        self.tokenizer.pre_tokenizer = pre_tokenizers.WhitespaceSplit()
        self.tokenizer.train_from_iterator(texts_bpe, trainer)
        self.tokenizer.add_tokens([newline_token])

        self.tokenizer.post_processor = TemplateProcessing(
            single=f"{SPECIAL_TOKENS['bos_token']} $A {SPECIAL_TOKENS['eos_token']}",
            special_tokens=[
                (SPECIAL_TOKENS['bos_token'], self.tokenizer.token_to_id(SPECIAL_TOKENS['bos_token'])),
                (SPECIAL_TOKENS['eos_token'], self.tokenizer.token_to_id(SPECIAL_TOKENS['eos_token'])),
            ],
        )
        self.tokenizer.decoder = WordPieceDecoder(prefix='##', cleanup=True)

        self.token2id = self.tokenizer.get_vocab()
        self.pad_token_id = self.token2id[SPECIAL_TOKENS['pad_token']]

    def encode(self, texts, max_length=None):
        if max_length is not None:
            self.tokenizer.enable_truncation(max_length=max_length)

        tokenized_texts = defaultdict(list)
        for text in texts:
            tokenized = self.tokenizer.encode(text.replace('\n', f' {newline_token} '))
            tokenized_texts['tokens'].append(tokenized.tokens)
            tokenized_texts['input_ids'].append(tokenized.ids)

        self.tokenizer.no_truncation()
        return tokenized_texts

    def decode(self, input_ids, skip_special_tokens=True):
        texts = []
        for ids in input_ids:
            texts.append(
                self.tokenizer.decode(
                    ids, skip_special_tokens=skip_special_tokens
                ).replace(self.newline_token, '\n').replace(' \n', '\n').replace('\n ', '\n')
            )
            
        return texts

    def __len__(self):
        return len(self.token2id)
```

### Обучение RNN

```python
import torch
import torch.nn as nn


class RNN(nn.Module):
    def __init__(self, vocab_size, hidden_size):
        super().__init__()
        self.hidden_size = hidden_size

        self.embeddings = nn.Embedding(vocab_size, hidden_size)

        self.W = nn.Linear(hidden_size + hidden_size, hidden_size)
        self.O = nn.Linear(hidden_size, vocab_size)

    def forward(self, input_ids, h0=None):
        batch_size, seq_len = input_ids.shape
        x = self.embeddings(input_ids)

        if h0 is None:
            h_t = torch.zeros(batch_size, self.hidden_size, device=x.device)
        else:
            h_t = h0

        outputs = []
        for t in range(seq_len):
            x_t = x[:, t, :]

            h_t = torch.tanh(
                self.W(torch.cat((x_t, h_t), dim=-1))
            )

            o_t = self.O(h_t)
            outputs.append(o_t)

        outputs = torch.stack(outputs, dim=1)
        return outputs, h_t
```

```python
def train(model, dataloader, optimizer, loss_fn):
    model.train()

    for input_ids in dataloader:
        input_ids = input_ids.to(device)

        logits, _ = model(input_ids)
        
        shift_ids = input_ids[:, 1:]
        shift_logits = logits[:, :-1]
        loss = loss_fn(shift_logits.permute(0, 2, 1), shift_ids)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        accuracy = (shift_logits.argmax(-1) == shift_ids).float().mean()
        # wandb.log({
        #     "train_loss": loss.item(),
        #     "train_accuracy": accuracy.item()
        # })


@torch.inference_mode()
def evaluate(model, dataloader, loss_fn):
    model.eval()

    accuracies = []
    losses = []
    for input_ids in dataloader:
        input_ids = input_ids.to(device)

        logits, _ = model(input_ids)
        
        shift_ids = input_ids[:, 1:]
        shift_logits = logits[:, :-1]
        loss = loss_fn(shift_logits.permute(0, 2, 1), shift_ids)

        accuracies.append((shift_logits.argmax(-1) == shift_ids).float().mean().item())
        losses.append(loss.item())

    loss = np.mean(losses)
    accuracy = np.mean(accuracies)
    
    # wandb.log({
    #     "test_loss": loss,
    #     "test_accuracy": accuracy
    # })

    return accuracy, loss
```

```python
tokenizer_class = WordTokenizer
tokenizer = tokenizer_class(corpus=train_texts, newline_token=newline_token, vocab_size=1024)
```

```python
from torch.nn.utils.rnn import pad_sequence

def collate_fn(batch, max_length=None):
    tokens = tokenizer.encode(batch, max_length=max_length)['input_ids']
    return pad_sequence(tokens, padding_value=tokenizer.pad_token_id, batch_first=True)
```

```python
from torch.utils.data import DataLoader
from functools import partial

max_length = 400
collate = partial(collate_fn, max_length=max_length)
train_loader = DataLoader(train_texts, collate_fn=collate, shuffle=True, batch_size=32)
test_loader = DataLoader(test_texts, collate_fn=collate, shuffle=False, batch_size=8)
```

```python
model = RNN(vocab_size=len(tokenizer), hidden_size=128)

optimizer = torch.optim.Adam(model.parameters(), lr=2e-4, weight_decay=1e-3)
loss_fn = nn.CrossEntropyLoss(ignore_index=tokenizer.pad_token_id)
```

```python
for epoch in range(5):
    train(model, train_loader, optimizer, loss_fn)
    test_accuracy, test_loss = evaluate(model, test_loader, loss_fn)
    print(test_accuracy, test_loss)
```
